```python
import pandas as pd
import glob
import os

# 基础路径（根据实际路径修改）
base_path = r"C:\Users\111\Desktop\武博卫星反演数据评估\RADI_CHN_MUL_HOR"

# 构建文件匹配模式（查找2024年所有符合命名规则的txt文件）
file_pattern = os.path.join(base_path, "**", "RADI_CHN_MUL_HOR_2024*.txt")
file_list = glob.glob(file_pattern, recursive=True)

print(f"找到 {len(file_list)} 个数据文件：")
for file in file_list:
    print(f"- {file}")

# 读取并合并所有文件
dfs = []
for file in file_list:
    try:
        # 提取文件名中的日期（例如从"RADI_CHN_MUL_HOR_2024100100.txt"中提取"20241001"）
        filename = os.path.basename(file)
        date_str = filename.split("_")[4][:8]  # 取第5个下划线后的前8位字符（年月日）
        
        # 读取空格分隔的txt文件（首行为列名）
        # sep='\s+'表示匹配任意数量的空格作为分隔符
        df = pd.read_csv(
            file,
            sep='\s+',  # 关键参数：处理空格分隔
            engine='python'  # 避免解析时的警告
        )
        
        # 添加日期列（转换为datetime格式，方便后续时间序列分析）
        df['date'] = pd.to_datetime(date_str, format='%Y%m%d')
        
        dfs.append(df)
        print(f"✅ 成功读取 {file}，数据行数：{len(df)}")
    
    except Exception as e:
        print(f"❌ 读取 {file} 失败：{str(e)}")

# 合并所有数据并输出结果
if dfs:
    combined_df = pd.concat(dfs, ignore_index=True)
    print(f"\n合并后的数据总量：{len(combined_df)} 行，{len(combined_df.columns)} 列")
    print("\n前5行数据预览：")
    print(combined_df.head())
    
    # 可选：保存合并后的数据为CSV（方便后续使用）
    output_path = os.path.join(base_path, "合并后的数据.csv")
    combined_df.to_csv(output_path, index=False)
    print(f"\n数据已保存至：{output_path}")
else:
    print("\n没有成功读取任何文件，请检查文件格式或路径是否正确。")
```

```python
import pandas as pd
import glob
import os
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
import tkinter as tk
from tkinter import ttk, messagebox
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg
from matplotlib.figure import Figure

# 设置中文字体显示
plt.rcParams["font.family"] = ["SimHei", "WenQuanYi Micro Hei", "Heiti TC"]
plt.rcParams["axes.unicode_minus"] = False  # 解决负号显示问题

# 读取数据函数
def read_data(base_path):
    # 构建文件匹配模式
    file_pattern = os.path.join(base_path, "**", "RADI_CHN_MUL_HOR_2024*.txt")
    file_list = glob.glob(file_pattern, recursive=True)
    
    print(f"找到 {len(file_list)} 个数据文件")
    dfs = []
    
    for file in file_list:
        try:
            filename = os.path.basename(file)
            date_str = filename.split("_")[4][:8]  # 提取日期
            time_str = filename.split("_")[4][8:10]  # 提取时间
            
            print(f"正在处理文件: {file}，日期: {date_str}，时间: {time_str}")
            
            # 验证日期和时间格式
            try:
                datetime_obj = datetime.strptime(date_str + time_str, "%Y%m%d%H")
            except ValueError as ve:
                print(f"⚠️ 日期时间格式错误: {ve}，跳过文件")
                continue
            
            # 读取空格分隔的txt文件
            df = pd.read_csv(file, sep='\s+', engine='python')
            
            # 添加日期列
            df['date'] = pd.to_datetime(date_str, format='%Y%m%d')
            
            # 修正：直接使用Timestamp的time()方法
            time_obj = pd.to_datetime(time_str, format='%H')
            df['time'] = time_obj.time()
            
            # 创建完整的日期时间列
            datetime_str = f"{date_str} {time_str}:00:00"
            df['datetime'] = pd.to_datetime(datetime_str, format='%Y%m%d %H:%M:%S')
            
            # 数据质量检查：替换无效值为NaN
            df = df.replace(999999, pd.NA)
            
            dfs.append(df)
            print(f"✅ 成功读取 {file}，行数: {len(df)}")
        except Exception as e:
            print(f"❌ 读取 {file} 失败: {str(e)}")
    
    if not dfs:
        messagebox.showerror("错误", "没有成功读取任何文件！")
        return None
    
    combined_df = pd.concat(dfs, ignore_index=True)
    print(f"合并后的数据量: {len(combined_df)} 行")
    
    # 保存合并后的数据
    output_path = os.path.join(base_path, "合并数据.csv")
    combined_df.to_csv(output_path, index=False)
    print(f"数据已保存至: {output_path}")
    
    return combined_df

# 绘制图表函数
def plot_parameter():
    # 获取用户选择
    station = station_var.get()
    parameter = parameter_var.get()
    
    if not station or not parameter:
        messagebox.showwarning("警告", "请选择站点和参数！")
        return
    
    # 筛选数据
    filtered_df = df[df['Station_Id_C'] == station].sort_values('datetime')
    
    if filtered_df.empty:
        messagebox.showwarning("警告", f"未找到站点 {station} 的数据！")
        return
    
    # 清除现有图表
    for widget in plot_frame.winfo_children():
        widget.destroy()
    
    # 创建图表
    fig = Figure(figsize=(12, 6), dpi=100)
    ax = fig.add_subplot(111)
    
    # 绘制参数随时间变化曲线
    sns.lineplot(x='datetime', y=parameter, data=filtered_df, ax=ax)
    
    # 设置图表标题和轴标签
    ax.set_title(f"站点 {station} 的 {parameter} 随时间变化趋势", fontsize=14)
    ax.set_xlabel("日期时间", fontsize=12)
    ax.set_ylabel(parameter, fontsize=12)
    
    # 添加网格线
    ax.grid(True, linestyle='--', alpha=0.7)
    
    # 自动旋转x轴标签以避免重叠
    fig.autofmt_xdate()
    
    # 添加数据点标记
    ax.plot(filtered_df['datetime'], filtered_df[parameter], 'o', markersize=4, alpha=0.6)
    
    # 将图表添加到Tkinter窗口
    canvas = FigureCanvasTkAgg(fig, master=plot_frame)
    canvas.draw()
    canvas.get_tk_widget().pack(fill=tk.BOTH, expand=True)

# 创建GUI界面
def create_gui(df):
    global station_var, parameter_var, plot_frame
    
    # 创建主窗口
    root = tk.Tk()
    root.title("气象数据可视化")
    root.geometry("1000x700")
    
    # 创建顶部选择区域
    control_frame = ttk.Frame(root, padding="10")
    control_frame.pack(fill=tk.X)
    
    # 站点选择
    ttk.Label(control_frame, text="选择站点:").grid(row=0, column=0, padx=5, pady=5, sticky=tk.W)
    stations = sorted(df['Station_Id_C'].unique())
    station_var = tk.StringVar()
    station_combobox = ttk.Combobox(control_frame, textvariable=station_var, values=stations, width=20)
    station_combobox.grid(row=0, column=1, padx=5, pady=5, sticky=tk.W)
    
    # 参数选择
    ttk.Label(control_frame, text="选择参数:").grid(row=0, column=2, padx=5, pady=5, sticky=tk.W)
    # 排除非数据列
    parameters = [col for col in df.columns if col not in ['Station_Id_C', 'Station_Id_d', 'Lat', 'Lon', 'Alti', 'date', 'time', 'datetime']]
    parameter_var = tk.StringVar()
    parameter_combobox = ttk.Combobox(control_frame, textvariable=parameter_var, values=parameters, width=30)
    parameter_combobox.grid(row=0, column=3, padx=5, pady=5, sticky=tk.W)
    
    # 绘制按钮
    plot_button = ttk.Button(control_frame, text="绘制图表", command=plot_parameter)
    plot_button.grid(row=0, column=4, padx=20, pady=5)
    
    # 创建图表显示区域
    plot_frame = ttk.Frame(root, padding="10")
    plot_frame.pack(fill=tk.BOTH, expand=True)
    
    # 添加初始提示
    ttk.Label(plot_frame, text="请选择站点和参数，然后点击'绘制图表'按钮").pack(pady=50)
    
    # 运行主循环
    root.mainloop()

# 主程序
if __name__ == "__main__":
    # 设置数据路径
    base_path = r"C:\Users\111\Desktop\武博卫星反演数据评估\RADI_CHN_MUL_HOR"
    
    # 读取数据
    df = read_data(base_path)
    print(df.head())  # 打印前几行数据以验证读取是否成功
    # if df is not None:
    #     # 创建GUI界面
    #     create_gui(df)
```

```python
import pandas as pd
import glob
import os
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
import numpy as np

# 设置中文字体显示
plt.rcParams["font.family"] = ["SimHei", "WenQuanYi Micro Hei", "Heiti TC"]
plt.rcParams["axes.unicode_minus"] = False  # 解决负号显示问题

# 要分析的站点和参数
TARGET_STATION = 50873
TARGET_PARAMETER = "V14316"

# 基础路径
base_path = r"C:\Users\111\Desktop\武博卫星反演数据评估\RADI_CHN_MUL_HOR"

# 使用glob递归查找所有txt文件
file_pattern = os.path.join(base_path, "**", "RADI_CHN_MUL_HOR_2024*.txt")
print(f"搜索模式: {file_pattern}")
file_list = glob.glob(file_pattern, recursive=True)
print(f"找到 {len(file_list)} 个数据文件")

# 读取并合并所有文件
dfs = []
for file in file_list:
    try:
        # 从文件路径解析日期信息
        filename = os.path.basename(file)
        date_str = filename.split('_')[4][:8]  # 提取20241001
        time_str = filename.split('_')[4][8:10]  # 提取小时部分
        print(f"正在处理文件: {file}，日期: {date_str}，时间: {time_str}")
        
        # 读取txt文件，使用空格作为分隔符
        df = pd.read_csv(file, sep='\s+')
        
        # 添加日期列
        df['date'] = pd.to_datetime(date_str, format='%Y%m%d')
        
        # 添加时间列
        time_obj = pd.to_datetime(time_str, format='%H')
        df['time'] = time_obj.time()
        
        # 创建完整的日期时间列
        datetime_str = f"{date_str} {time_str}:00:00"
        df['datetime'] = pd.to_datetime(datetime_str, format='%Y%m%d %H:%M:%S')
        
        dfs.append(df)
        print(f"成功读取 {len(df)} 行数据")
        
    except Exception as e:
        print(f"读取文件 {file} 失败: {str(e)}")

# 合并所有数据
if dfs:
    combined_df = pd.concat(dfs, ignore_index=True)
    print(f"合并后的数据量: {len(combined_df)} 行")
    print(combined_df.head())  # 打印前几行数据以验证读取是否成功
    data = combined_df.copy()[['Station_Id_C', 'Station_Id_d', 'Lat', 'Lon', 'Alti', 'date', 'time', 'datetime'] + [col for col in combined_df.columns if col not in ['Station_Id_C', 'Station_Id_d', 'Lat', 'Lon', 'Alti', 'date', 'time', 'datetime']]]
    # 筛选目标站点的数据
    station_data = combined_df[combined_df['Station_Id_C'] == TARGET_STATION].copy()
    
    if not station_data.empty:
        print(f"找到站点 {TARGET_STATION} 的 {len(station_data)} 条数据")
        
        # 检查目标参数是否存在
        if TARGET_PARAMETER in station_data.columns:
            # 处理缺失值（将999999替换为NaN）
            station_data[TARGET_PARAMETER] = station_data[TARGET_PARAMETER].replace(999999, pd.NA)
            
            # 移除目标参数列中所有的NaN值
            valid_data = station_data.dropna(subset=[TARGET_PARAMETER])
            
            if valid_data.empty:
                print(f"错误: 站点 {TARGET_STATION} 的 {TARGET_PARAMETER} 参数全部为缺失值")
            else:
                print(f"有效数据点: {len(valid_data)}/{len(station_data)}")
                
                # 按日期时间排序
                valid_data = valid_data.sort_values('datetime')
                
                # 数据预览
                print("\n数据预览:")
                display(valid_data[['datetime', TARGET_PARAMETER]].head())
                
                # 计算统计信息
                stats = valid_data[TARGET_PARAMETER].describe()
                print(f"\n{TARGET_PARAMETER} 参数统计摘要:")
                display(stats)
                
                # 创建图表
                plt.figure(figsize=(14, 7))
                
                # 绘制随时间变化曲线
                sns.lineplot(x='datetime', y=TARGET_PARAMETER, data=valid_data)
                
                # 添加数据点标记
                plt.plot(valid_data['datetime'], valid_data[TARGET_PARAMETER], 'o', markersize=6, alpha=0.6)
                
                # 设置图表标题和轴标签
                plt.title(f"站点 {TARGET_STATION} 的 {TARGET_PARAMETER} 参数随时间变化", fontsize=16)
                plt.xlabel("日期时间", fontsize=14)
                plt.ylabel(f"{TARGET_PARAMETER} 值", fontsize=14)
                
                # 添加网格线
                plt.grid(True, linestyle='--', alpha=0.7)
                
                # 自动旋转x轴标签以避免重叠
                plt.xticks(rotation=45)
                
                # 显示图表
                plt.tight_layout()
                plt.show()
                
        else:
            print(f"错误: 数据中不存在参数 {TARGET_PARAMETER}")
            print(f"可用参数: {', '.join(combined_df.columns.tolist())}")
    else:
        print(f"错误: 未找到站点 {TARGET_STATION} 的数据")
else:
    print("没有成功读取任何文件")
```
