```python
import os
import re
import pandas as pd
import numpy as np
import rasterio
from datetime import datetime, timedelta
from sklearn.metrics import mean_squared_error, mean_absolute_error
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from matplotlib import font_manager

# 设置中文显示 - 更健壮的字体设置方案
try:
    # 尝试使用系统自带的中文字体
    font_path = font_manager.findSystemFonts(fontpaths=None, fontext='ttf')
    chinese_fonts = [f for f in font_path if any(name in f.lower() for name in ['simhei', 'msyh', 'simsun', 'arial unicode'])]
    
    if chinese_fonts:
        # 优先使用黑体，其次雅黑，再其次宋体
        for preferred in ['simhei', 'msyh', 'simsun']:
            selected = [f for f in chinese_fonts if preferred in f.lower()]
            if selected:
                font_prop = font_manager.FontProperties(fname=selected[0])
                plt.rcParams['font.family'] = font_prop.get_name()
                break
    else:
        # 如果没有找到任何中文字体，使用默认字体并警告
        print("警告: 未找到系统中文字体，中文显示可能不正常")
except Exception as e:
    print(f"字体设置出错: {e}")

# 确保负号显示正常
plt.rcParams["axes.unicode_minus"] = False

# 定义单位常量
UNIT = "W/m²"
UNIT_LABEL = f"辐射值 ({UNIT})"
UNIT_SHORT = "W m$^{-2}$"

def read_obs_data(file_path):
    """读取地面观测CSV数据"""
    try:
        obs_df = pd.read_csv(
            file_path, 
            encoding='gbk', 
            parse_dates=['Datetime'], 
            date_parser=lambda x: pd.to_datetime(x, errors='coerce')
        )
        # 数据清洗
        obs_df = obs_df.dropna(subset=['Datetime'])
        obs_df = obs_df[obs_df['radi'] >= 0]
        # 筛选11-12月数据
        obs_df = obs_df[(obs_df['Datetime'].dt.month >= 11) & (obs_df['Datetime'].dt.month <= 12)]
        # 小时重采样
        obs_df = obs_df.set_index('Datetime').resample('h').mean().reset_index()
        obs_df = obs_df.dropna(subset=['radi'])
        return obs_df
    except Exception as e:
        print(f"读取地面观测数据失败: {e}")
        return pd.DataFrame(columns=['Datetime', 'radi'])

def find_satellite_files(base_dir, pattern, time_range=(11, 12)):
    """查找卫星TIFF文件"""
    file_list = []
    date_pattern = r'(\d{4})(\d{2})(\d{2})(\d{2})(\d{2})(\d{2})'
    for root, dirs, files in os.walk(base_dir):
        for file in files:
            if pattern in file and file.lower().endswith(('.tif', '.tiff')):
                match = re.search(date_pattern, file)
                if match and time_range[0] <= int(match.group(2)) <= time_range[1]:
                    file_list.append(os.path.join(root, file))
    return file_list

def extract_pixel_value(tif_path, lon, lat, nodata=-9999):
    """提取TIFF中指定点的辐射值"""
    try:
        with rasterio.open(tif_path) as src:
            if not (src.bounds.left <= lon <= src.bounds.right and src.bounds.bottom <= lat <= src.bounds.top):
                print(f"坐标({lon}, {lat})不在影像范围内: {os.path.basename(tif_path)}")
                return np.nan
            row, col = src.index(lon, lat)
            data = src.read(1)
            if 0 <= row < data.shape[0] and 0 <= col < data.shape[1]:
                pixel_value = data[row, col]
                if pixel_value in (nodata, np.nan) or not (0 <= pixel_value <= 1500):
                    return np.nan
                return pixel_value
            return np.nan
    except Exception as e:
        print(f"处理TIFF出错 {os.path.basename(tif_path)}: {e}")
        return np.nan

def match_satellite_to_obs(obs_df, satellite_files, station_lon, station_lat, file_type='uncorrected'):
    """匹配卫星与地面数据，并输出对齐信息"""
    result_df = pd.DataFrame(columns=['Datetime', 'radi', f'{file_type}_value'])
    if obs_df.empty:
        print("地面数据为空，无法匹配")
        return result_df
    
    date_pattern = r'(\d{4})(\d{2})(\d{2})(\d{2})(\d{2})(\d{2})'
    matched_pairs = []
    
    for sat_file in satellite_files:
        match = re.search(date_pattern, os.path.basename(sat_file))
        if not match:
            continue
        
        try:
            sat_time_utc = datetime(*map(int, match.groups()))
            sat_time_bjt = sat_time_utc + timedelta(hours=8)
        except ValueError:
            continue
        
        time_diff = (obs_df['Datetime'] - sat_time_bjt).dt.total_seconds().abs()
        if time_diff.min() > 1800:
            continue
        
        closest_idx = time_diff.idxmin()
        ground_time = obs_df.loc[closest_idx, 'Datetime']
        pixel_value = extract_pixel_value(sat_file, station_lon, station_lat)
        
        if not np.isnan(pixel_value):
            matched_pairs.append({
                '卫星文件': os.path.basename(sat_file),
                '卫星时间(UTC)': sat_time_utc,
                '卫星时间(BJT)': sat_time_bjt,
                '地面观测时间': ground_time,
                '时间差(秒)': time_diff.min(),
                '辐射值': pixel_value
            })
            
            result_df = pd.concat([result_df, pd.DataFrame({
                'Datetime': [ground_time],
                'radi': [obs_df.loc[closest_idx, 'radi']],
                f'{file_type}_value': [pixel_value]
            })], ignore_index=True)
    
    # 输出匹配结果
    if matched_pairs:
        match_df = pd.DataFrame(matched_pairs)
        print(f"\n{'='*50}")
        print(f"{file_type}数据匹配详情（共{match_df.shape[0]}对）")
        print("="*50)
        print(match_df[['卫星文件', '卫星时间(BJT)', '地面观测时间', '时间差(秒)', '辐射值']])
        
        # 可视化时间对齐情况
        plt.figure(figsize=(12, 6))
        plt.plot(match_df['卫星时间(BJT)'], match_df['时间差(秒)']/60, 'ro', label='时间差(分钟)')
        plt.axhline(y=30, color='gray', linestyle='--', label='30分钟阈值')
        plt.title(f"{file_type}数据时间对齐情况\n(最大时间差: {match_df['时间差(秒)'].max()/60:.1f}分钟)", fontsize=12)
        plt.xlabel('卫星过境时间(BJT)', fontsize=10)
        plt.ylabel('与地面观测的时间差(分钟)', fontsize=10)
        plt.legend()
        plt.grid(linestyle='--', alpha=0.5)
        plt.tight_layout()
        plt.show()
    else:
        print(f"警告: 未找到有效的{file_type}匹配数据")
    
    return result_df.drop_duplicates(subset=['Datetime'])

def calculate_metrics(obs_values, pred_values):
    """计算误差指标"""
    valid = ~(np.isnan(obs_values) | np.isnan(pred_values))
    obs, pred = obs_values[valid], pred_values[valid]
    if len(obs) == 0:
        return np.nan, np.nan, 0
    rmse = np.sqrt(mean_squared_error(obs, pred))
    mae = mean_absolute_error(obs, pred)
    return rmse, mae, len(obs)

def visualize_comparison(obs_df, title="辐射值对比"):
    """可视化对比结果"""
    plt.figure(figsize=(15, 8))
    font_props = {'family': plt.rcParams['font.family']} if 'font.family' in plt.rcParams else {}
    
    plt.plot(obs_df['Datetime'], obs_df['radi'], 'k-', marker='.', label='地面观测', alpha=0.8)
    if 'uncorrected_value' in obs_df.columns:
        plt.plot(obs_df['Datetime'], obs_df['uncorrected_value'], 'r-', marker='^', label='订正前卫星', alpha=0.7)
    if 'corrected_value' in obs_df.columns:
        plt.plot(obs_df['Datetime'], obs_df['corrected_value'], 'b-', marker='s', label='订正后卫星', alpha=0.7)
    
    plt.title(f"{title} (单位: {UNIT_SHORT})", fontsize=14, **font_props)
    plt.xlabel('日期时间', fontsize=12, **font_props)
    plt.ylabel(f"辐射值 ({UNIT_SHORT})", fontsize=12, **font_props)
    
    plt.grid(linestyle='--', alpha=0.5)
    plt.legend(prop=font_props)
    plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%m-%d %H:%M'))
    plt.gcf().autofmt_xdate()
    
    # 添加统计信息
    stats_text = []
    if 'uncorrected_value' in obs_df.columns:
        rmse_uncorr, mae_uncorr, n_uncorr = calculate_metrics(obs_df['radi'], obs_df['uncorrected_value'])
        stats_text.append(f"订正前: RMSE={rmse_uncorr:.2f} {UNIT_SHORT}, MAE={mae_uncorr:.2f} {UNIT_SHORT}")
    if 'corrected_value' in obs_df.columns:
        rmse_corr, mae_corr, n_corr = calculate_metrics(obs_df['radi'], obs_df['corrected_value'])
        stats_text.append(f"订正后: RMSE={rmse_corr:.2f} {UNIT_SHORT}, MAE={mae_corr:.2f} {UNIT_SHORT}")
    
    if stats_text:
        plt.figtext(0.02, 0.02, "\n".join(stats_text), 
                    bbox=dict(facecolor='white', alpha=0.8), **font_props)
    
    return plt

def main():
    # 路径配置
    obs_file = r"C:\Users\111\Desktop\qihui_obs_update.csv"
    satellite_dir = r"C:\Users\111\Desktop\武博卫星反演数据评估\FY4B_SSR_corrected"
    station_lon, station_lat = 100.51, 36.02

    # 读取地面数据
    obs_df = read_obs_data(obs_file)
    print(f"\n地面数据有效点数: {len(obs_df)}")
    if len(obs_df) == 0:
        return

    # 处理卫星数据
    uncorrected_files = find_satellite_files(satellite_dir, "MSP3_PMSC_ENGSOLMFC_SSI_CHN_4KM_")
    corrected_files = find_satellite_files(satellite_dir, "MSP3_PMSC_ENGSOLMFC_SSI-correct_CHN_4KM_")
    print(f"\n订正前卫星文件数: {len(uncorrected_files)}")
    print(f"订正后卫星文件数: {len(corrected_files)}")

    # 匹配数据（将显示详细对齐信息）
    print("\n开始匹配订正前数据...")
    uncorrected = match_satellite_to_obs(obs_df, uncorrected_files, station_lon, station_lat, 'uncorrected')
    
    print("\n开始匹配订正后数据...")
    corrected = match_satellite_to_obs(obs_df, corrected_files, station_lon, station_lat, 'corrected')
    
    # 合并结果
    merged = pd.merge(uncorrected, corrected[['Datetime', 'corrected_value']], on='Datetime', how='outer')
    print(f"\n最终有效匹配点数 - 订正前: {len(uncorrected)}, 订正后: {len(corrected)}")

    # 计算指标
    rmse_uncorr, mae_uncorr, n_uncorr = calculate_metrics(merged['radi'], merged['uncorrected_value'])
    rmse_corr, mae_corr, n_corr = calculate_metrics(merged['radi'], merged['corrected_value'])

    # 输出结果
    print(f"\n精度对比（单位: {UNIT}）:")
    print(f"订正前 - RMSE: {rmse_uncorr:.2f}, MAE: {mae_uncorr:.2f}, 样本数: {n_uncorr}")
    print(f"订正后 - RMSE: {rmse_corr:.2f}, MAE: {mae_corr:.2f}, 样本数: {n_corr}")

    # 可视化
    plt = visualize_comparison(merged)
    plt.savefig(r"C:\Users\111\Desktop\辐射对比.png", dpi=300, bbox_inches='tight')
    plt.show()

if __name__ == "__main__":
    main()
```
